```python
import cv2
```

```python
bgr_image = cv2.imread("images/butterfly.jpg")
```

```python
gray_image = cv2.cvtColor(bgr_image, cv2.COLOR_BGR2GRAY)

cv2.imwrite("images/butterfly_gray.jpg", gray_image)
```

```python
hsv_image = cv2.cvtColor(bgr_image, cv2.COLOR_BGR2HSV)

cv2.imwrite("images/butterfly_hsv.jpg", hsv_image)
```

```python
lab_image = cv2.cvtColor(bgr_image, cv2.COLOR_BGR2Lab)

cv2.imwrite("images/butterfly_lab.jpg", lab_image)
```

```python
rgb_image = cv2.cvtColor(bgr_image, cv2.COLOR_BGR2RGB)

cv2.imwrite("images/butterfly_rgb.jpg", rgb_image)
```
