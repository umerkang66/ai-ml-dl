```python
words = [
    "eat", "eating", "eats", "eaten", "ate",
    "run", "running", "runs", "runner",
    "play", "playing", "played", "player", "plays",
    "study", "studies", "studying", "studied",
    "connect", "connected", "connecting", "connection", "connections",
    "happy", "happiness", "happier", "happiest",
    "compute", "computer", "computing", "computation", "computes",
    "argue", "arguing", "argued", "argument", "arguably",
    "fly", "flies", "flying", "flew",
    "organize", "organizing", "organized", "organization"
]
```

## PorterStemmer


```python
from nltk.stem import PorterStemmer, RegexpStemmer, SnowballStemmer
```

```python
porter_stemmer = PorterStemmer()

for word in words:
    stemmed_word = porter_stemmer.stem(word)
    print(f"Original: {word}, Stemmed: {stemmed_word}")
```

```python
porter_stemmer.stem('congratulations')
```

```python
porter_stemmer.stem("ssitting")
```

## Regexp Stemmer


```python
regexp_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
regexp_stemmer.stem('eating')
```

```python
for word in words:
    stemmed_word = regexp_stemmer.stem(word)
    print(f"Original: {word}, Stemmed: {stemmed_word}")
```

## Snowbell Stemmer


```python
snowball_stemmer = SnowballStemmer("english")
```

```python
for word in words:
    stemmed_word = snowball_stemmer.stem(word)
    print(f"Original: {word}, Stemmed: {stemmed_word}")
```

```python
porter_stemmer.stem('fairly'), porter_stemmer.stem('sportingly')
```

```python
snowball_stemmer.stem('fairly'), snowball_stemmer.stem('sportingly')
```
