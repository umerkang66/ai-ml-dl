```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
dataset = pd.read_csv("Algerian_forest_fires_dataset_UPDATE.csv", header=1)
```

```python
dataset.head()
```

```python
dataset.info()
```

## Data Cleaning


```python
# dataset.isnull().sum()
dataset[dataset.isnull().any(axis=1)]
```

```python
# bejaia dataset will have index 0, other will have index 1

dataset.loc[:122, "Region"] = 0
dataset.loc[122:, "Region"] = 1
```

```python
dataset[["Region"]] = dataset[["Region"]].astype(int)
```

```python
dataset.head()
```

```python
dataset.isnull().sum()
```

```python
dataset = dataset.dropna().reset_index(drop=True)
```

```python
dataset.isnull().sum()
```

```python
dataset = dataset.drop(122).reset_index(drop=True)
```

```python
dataset.iloc[[122]]
```

```python
dataset.columns
```

```python
# fix spaces in column names
dataset.columns = dataset.columns.str.strip()
```

```python
dataset.columns
```

```python
dataset.head()
```

```python
dataset.info()
```

```python
dataset[["month", "day", "year", "Temperature", "RH", "Ws"]] = dataset[
    ["month", "day", "year", "Temperature", "RH", "Ws"]
].astype(int)
```

```python
dataset.info()
```

```python
# Changing the other columns to float data type

objects = [
    features for features in dataset.columns if dataset[features].dtype == "object"
]
```

```python
for i in objects:
    if i != "Classes":
        dataset[i] = dataset[i].astype(float)
```

```python
dataset.info()
```

```python
dataset.describe()
```

```python
# Let's save the cleaned dataset
dataset.to_csv("algeria_forest_fires_cleaned.csv", index=False)
```

## Exploratory Data Analysis


```python
df_copy = dataset.drop(["day", "month", "year"], axis=1)
```

```python
df_copy.head()
```

```python
# Encoding of the categories in the classes
df_copy["Classes"] = np.where(df_copy["Classes"].str.contains("not fire"), 0, 1)
```

```python
df_copy.head()
```

```python
df_copy.tail()
```

```python
# Plot density plots for all features

df_copy.hist(figsize=(15, 10), bins=50)
plt.show()
```

```python
percentage = df_copy["Classes"].value_counts(normalize=True) * 100
```

```python
# plotting pi chart
classlabels = ["Fire", "Not Fire"]
plt.figure(figsize=(12, 7))
plt.pie(percentage, labels=classlabels, autopct="%1.1f%%")
plt.title("Distribution of Fire and Not Fire Classes")
plt.show()
```

```python
df_copy.corr()
```

```python
sns.heatmap(df_copy.corr())
plt.show()
```

```python
# Encoding of the categories in the classes
dataset["Classes"] = np.where(
    dataset["Classes"].str.contains("not fire"), "not fire", "fire"
)
```

```python
# Monthly fire analysis

dftemp = dataset.loc[dataset["Region"] == 0]
plt.subplots(figsize=(13, 6))

sns.set_style("whitegrid")
sns.countplot(x="month", hue="Classes", data=dftemp)
plt.ylabel("No of Fires", weight="bold")
plt.xlabel("Month", weight="bold")
plt.title("Fire Analysis of Bejaia Region", weight="bold")
plt.show()
```

```python
# Monthly fire analysis

dftemp = dataset.loc[dataset["Region"] == 1]
plt.subplots(figsize=(13, 6))

sns.set_style("whitegrid")
sns.countplot(x="month", hue="Classes", data=dftemp)
plt.ylabel("No of Fires", weight="bold")
plt.xlabel("Month", weight="bold")
plt.title("Fire Analysis of Sidi - Bell Region", weight="bold")
plt.show()
```
