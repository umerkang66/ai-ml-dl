```python
import tensorflow as tf
import pickle
import pandas as pd
import numpy as np
```

# Load the pickle files


```python
model = tf.keras.models.load_model("./models/churn_model.h5")

with open("./models/gender_label_encoder.pkl", "rb") as f:
    gender_label_encoder = pickle.load(f)

with open("./models/one_hot_geography_encoder.pkl", "rb") as f:
    one_hot_geography_encoder = pickle.load(f)

with open("./models/scaler.pkl", "rb") as f:
    scaler = pickle.load(f)
```

```python
input_data = {
    "CreditScore": 600,
    "Geography": "France",
    "Gender": "Male",
    "Age": 40,
    "Tenure": 3,
    "Balance": 60000,
    "NumOfProducts": 2,
    "HasCrCard": 1,
    "IsActiveMember": 1,
    "EstimatedSalary": 50000,
}
```

```python
df = pd.DataFrame([input_data])
```

```python
df.head()
```

```python
df["Gender"] = gender_label_encoder.transform(df["Gender"])


df = pd.concat(
    [
        df.drop("Geography", axis=1),
        pd.DataFrame(
            one_hot_geography_encoder.transform(df[["Geography"]]).toarray(),
            columns=one_hot_geography_encoder.get_feature_names_out(["Geography"]),
        ),
    ],
    axis=1,
)

X = scaler.transform(df)
```

```python
model.predict(X)
```

```python
if model.predict(X)[0][0] > 0.5:
    print("The customer is likely to leave (churn) the bank.")
else:
    print("The customer is likely to stay with the bank.")
```
